```python
import os
import zipfile
import shutil
import random
import glob
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import seaborn as sns
from torchvision import datasets, transforms
from torchvision.models import resnet50, ResNet50_Weights
from sklearn.metrics import (classification_report, confusion_matrix, accuracy_score, precision_score, recall_score, f1_score)

# Reproducibility

def set_seed(seed: int = 42):
    random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
use_amp = torch.cuda.is_available()
print("Using device:", device, "| AMP:", use_amp)

# Unzip dataset

try:
    from google.colab import drive
    drive.mount("/content/drive")
except Exception:
    pass

zip_path = "/content/drive/MyDrive/DLBD/bird_CUB_200_2011.zip"
output_dir = "/content/CUB_200_2011"

if os.path.exists(output_dir):
    print(f"Removing existing dataset directory: {output_dir}")
    shutil.rmtree(output_dir)

os.makedirs(output_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zf:
    zf.extractall(output_dir)
print(f"Unzipped {zip_path} -> {output_dir}")

DATASET_ROOT = os.path.join(output_dir, "CUB_200_2011")
images_txt   = os.path.join(DATASET_ROOT, "images.txt")
split_txt    = os.path.join(DATASET_ROOT, "train_test_split.txt")
images_dir   = os.path.join(DATASET_ROOT, "images")

BASE_SPLIT_DIR = os.path.join(DATASET_ROOT, "data_split")
train_dir = os.path.join(BASE_SPLIT_DIR, "train")
val_dir   = os.path.join(BASE_SPLIT_DIR, "val")
test_dir  = os.path.join(BASE_SPLIT_DIR, "test")

os.makedirs(train_dir, exist_ok=True)
os.makedirs(val_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

# Build official train/test split
def build_official_split():
    id_to_path = {}
    with open(images_txt, "r") as f:
        for line in f:
            img_id, rel_path = line.strip().split(" ", 1)
            id_to_path[int(img_id)] = rel_path

    id_to_split = {}
    with open(split_txt, "r") as f:
        for line in f:
            img_id, is_train = line.strip().split()
            id_to_split[int(img_id)] = int(is_train)

    train_count = 0
    test_count = 0
    skipped_existing = 0

    for img_id in sorted(id_to_path.keys()):
        rel_path = id_to_path[img_id]
        src = os.path.join(images_dir, rel_path)

        if id_to_split[img_id] == 1:
            dst = os.path.join(train_dir, rel_path)
            train_count += 1
        else:
            dst = os.path.join(test_dir, rel_path)
            test_count += 1

        os.makedirs(os.path.dirname(dst), exist_ok=True)

        if os.path.exists(dst):
            skipped_existing += 1
            continue

        shutil.copy2(src, dst)

    print("Official split ready")
    print("Train images:", train_count)
    print("Test images :", test_count)
    print("Skipped existing:", skipped_existing)

build_official_split()

# Deterministic train val split
def make_val_split(split_ratio=0.1, seed=42):
    rng = random.Random(seed)

    existing_val = glob.glob(os.path.join(val_dir, "**", "*.jpg"), recursive=True)
    if len(existing_val) > 0:
        print(f"Val split already exists ({len(existing_val)} images). Skipping move.")
        return

    class_names = [d for d in sorted(os.listdir(train_dir)) if os.path.isdir(os.path.join(train_dir, d))]

    moved = 0
    for class_name in class_names:
        class_path = os.path.join(train_dir, class_name)
        images = [f for f in sorted(os.listdir(class_path)) if f.lower().endswith(".jpg")]

        rng.shuffle(images)
        val_count = int(len(images) * split_ratio)
        val_images = images[:val_count]

        dst_class = os.path.join(val_dir, class_name)
        os.makedirs(dst_class, exist_ok=True)

        for img in val_images:
            src = os.path.join(class_path, img)
            dst = os.path.join(dst_class, img)
            shutil.move(src, dst)
            moved += 1

    print(f"Validation split created (moved {moved} images).")

make_val_split(split_ratio=0.1, seed=42)

train_count = len(glob.glob(os.path.join(train_dir, "**", "*.jpg"), recursive=True))
val_count   = len(glob.glob(os.path.join(val_dir, "**", "*.jpg"), recursive=True))
test_count  = len(glob.glob(os.path.join(test_dir, "**", "*.jpg"), recursive=True))
print(f"Counts | train={train_count} val={val_count} test={test_count}")

# Datasets & loaders

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

std_train_tfms = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomResizedCrop(224, scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

std_eval_tfms = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

custom_train_tfms = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomResizedCrop(224, scale=(0.6, 1.0), ratio=(0.75, 1.33)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    transforms.RandomErasing(p=0.25, scale=(0.02, 0.12), ratio=(0.3, 3.3), value="random"),
])

std_train_ds = datasets.ImageFolder(train_dir, transform=std_train_tfms)
std_val_ds   = datasets.ImageFolder(val_dir,   transform=std_eval_tfms)
std_test_ds  = datasets.ImageFolder(test_dir,  transform=std_eval_tfms)

custom_train_ds = datasets.ImageFolder(train_dir, transform=custom_train_tfms)
custom_val_ds   = datasets.ImageFolder(val_dir,   transform=std_eval_tfms)
custom_test_ds  = datasets.ImageFolder(test_dir,  transform=std_eval_tfms)

num_classes = len(std_train_ds.classes)
print("Num classes:", num_classes)

batch_size = 32
num_workers = 2
pin_memory = torch.cuda.is_available()

g = torch.Generator()
g.manual_seed(42)

def seed_worker(worker_id):
    worker_seed = 42 + worker_id
    np.random.seed(worker_seed)
    random.seed(worker_seed)

std_train_loader = DataLoader(std_train_ds, batch_size=batch_size, shuffle=True,
                              num_workers=num_workers, pin_memory=pin_memory,
                              worker_init_fn=seed_worker, generator=g)
std_val_loader   = DataLoader(std_val_ds,   batch_size=batch_size, shuffle=False,
                              num_workers=num_workers, pin_memory=pin_memory)
std_test_loader  = DataLoader(std_test_ds,  batch_size=batch_size, shuffle=False,
                              num_workers=num_workers, pin_memory=pin_memory)

custom_train_loader = DataLoader(custom_train_ds, batch_size=batch_size, shuffle=True,
                                 num_workers=num_workers, pin_memory=pin_memory,
                                 worker_init_fn=seed_worker, generator=g)
custom_val_loader   = DataLoader(custom_val_ds,   batch_size=batch_size, shuffle=False,
                                 num_workers=num_workers, pin_memory=pin_memory)
custom_test_loader  = DataLoader(custom_test_ds,  batch_size=batch_size, shuffle=False,
                                 num_workers=num_workers, pin_memory=pin_memory)

# Metrics

@torch.no_grad()
def predict_all(model, loader):
    model.eval()
    y_true, y_pred = [], []
    for x, y in loader:
        x = x.to(device, non_blocking=True)
        logits = model(x)
        preds = logits.argmax(1).cpu().numpy()
        y_true.extend(y.numpy().tolist())
        y_pred.extend(preds.tolist())
    return np.array(y_true), np.array(y_pred)

def compute_metrics(y_true, y_pred):
    return {
        "accuracy":  accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, average="macro", zero_division=0),
        "recall":    recall_score(y_true, y_pred, average="macro", zero_division=0),
        "f1":        f1_score(y_true, y_pred, average="macro", zero_division=0),
    }

def plot_confusion_matrix(y_true, y_pred, title="Confusion Matrix"):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, cmap="viridis")
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.show()

def plot_learning_curves(history, title_prefix=""):
    epochs = np.arange(1, len(history["train_loss"]) + 1)

    # Loss
    plt.figure(figsize=(8, 4))
    plt.plot(epochs, history["train_loss"], label="train loss")
    plt.plot(epochs, history["val_loss"], label="val loss")
    plt.title(f"{title_prefix} Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

    # Accuracy
    plt.figure(figsize=(8, 4))
    plt.plot(epochs, history["train_acc"], label="train acc")
    plt.plot(epochs, history["val_acc"], label="val acc")
    plt.title(f"{title_prefix} Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()

def plot_normalized_confusion_matrix(y_true, y_pred, title="Normalized Confusion Matrix"):
    cm = confusion_matrix(y_true, y_pred)
    cm = cm.astype(np.float32) / np.clip(cm.sum(axis=1, keepdims=True), 1, None)

    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, cmap="viridis")
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.show()

# Shared train/evaluation loops

def train_one_epoch(model, loader, optimizer, criterion, scaler):
    model.train()
    total_loss = 0.0
    total_correct = 0
    total = 0

    for x, y in loader:
        x = x.to(device, non_blocking=True)
        y = y.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        with torch.cuda.amp.autocast(enabled=use_amp):
            logits = model(x)
            loss = criterion(logits, y)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item() * x.size(0)
        total_correct += (logits.argmax(1) == y).sum().item()
        total += x.size(0)

    return total_loss / total, total_correct / total

@torch.no_grad()
def eval_one_epoch(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    total_correct = 0
    total = 0

    for x, y in loader:
        x = x.to(device, non_blocking=True)
        y = y.to(device, non_blocking=True)

        logits = model(x)
        loss = criterion(logits, y)

        total_loss += loss.item() * x.size(0)
        total_correct += (logits.argmax(1) == y).sum().item()
        total += x.size(0)

    return total_loss / total, total_correct / total

# Model 1: Pretrained ResNet50

def build_resnet50(num_classes: int, weights=None):
    model = resnet50(weights=weights)
    model.fc = nn.Linear(model.fc.in_features, num_classes)
    return model.to(device)

def finetune_resnet50():
    model = build_resnet50(num_classes, weights=ResNet50_Weights.DEFAULT)
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

    best_val_acc = 0.0
    best_path = "/content/best_resnet50_cub.pth"

    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}

    # Phase 1: train head only
    for name, p in model.named_parameters():
        p.requires_grad = name.startswith("fc.")

    optimizer = optim.AdamW([p for p in model.parameters() if p.requires_grad],
                            lr=1e-3, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=5)
    scaler = torch.cuda.amp.GradScaler(enabled=use_amp)

    phase1_epochs = 5

    print("\n=ResNet50 Phase 1 (train head only) ===")
    for epoch in range(1, phase1_epochs + 1):
        tr_loss, tr_acc = train_one_epoch(model, std_train_loader, optimizer, criterion, scaler)
        va_loss, va_acc = eval_one_epoch(model, std_val_loader, criterion)
        scheduler.step()

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(va_loss)
        history["val_acc"].append(va_acc)

        print(f"[P1] Epoch {epoch:02d}/{phase1_epochs} | train acc {tr_acc:.4f} | val acc {va_acc:.4f}")

        if va_acc > best_val_acc:
            best_val_acc = va_acc
            torch.save({"model_state": model.state_dict(),
                        "class_to_idx": std_train_ds.class_to_idx,
                        "val_acc": best_val_acc}, best_path)

    # Phase 2: fine-tune all
    for p in model.parameters():
        p.requires_grad = True

    optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=25)
    scaler = torch.cuda.amp.GradScaler(enabled=use_amp)

    phase2_epochs = 25

    print("\nResNet50 Phase 2")
    for epoch in range(1, phase2_epochs + 1):
        tr_loss, tr_acc = train_one_epoch(model, std_train_loader, optimizer, criterion, scaler)
        va_loss, va_acc = eval_one_epoch(model, std_val_loader, criterion)
        scheduler.step()

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(va_loss)
        history["val_acc"].append(va_acc)

        print(f"[P2] Epoch {epoch:02d}/{phase2_epochs} | train acc {tr_acc:.4f} | val acc {va_acc:.4f}")

        if va_acc > best_val_acc:
            best_val_acc = va_acc
            torch.save({"model_state": model.state_dict(),
                        "class_to_idx": std_train_ds.class_to_idx,
                        "val_acc": best_val_acc}, best_path)
            print(f"Saved new best ResNet50 checkpoint: val_acc={best_val_acc:.4f}")

    print("Best ResNet50 val acc:", best_val_acc)
    print("Best ResNet50 ckpt:", best_path)

    plot_learning_curves(history, title_prefix="ResNet50")

    return best_path

# Model 2: CustomCNN
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=8):
        super().__init__()
        self.pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction),
            nn.ReLU(),
            nn.Linear(channels // reduction, channels),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.shape
        s = self.pool(x).view(b, c)
        s = self.fc(s).view(b, c, 1, 1)
        return x * s

class MultiBranchBlock(nn.Module):
    def __init__(self, in_ch, out_ch, stride=1):
        super().__init__()
        b1 = out_ch // 4
        b2 = out_ch // 4
        b3 = out_ch // 4
        b4 = out_ch - (b1 + b2 + b3)

        self.branch1 = nn.Sequential(
            nn.Conv2d(in_ch, b1, kernel_size=1, stride=stride, bias=False),
            nn.BatchNorm2d(b1),
            nn.ReLU(),
        )
        self.branch2 = nn.Sequential(
            nn.Conv2d(in_ch, b2, kernel_size=3, stride=stride, padding=1, bias=False),
            nn.BatchNorm2d(b2),
            nn.ReLU(),
        )
        self.branch3 = nn.Sequential(
            nn.Conv2d(in_ch, b3, kernel_size=3, stride=stride, padding=1, bias=False),
            nn.BatchNorm2d(b3),
            nn.ReLU(),
            nn.Conv2d(b3, b3, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(b3),
            nn.ReLU(),
        )
        self.branch4 = nn.Sequential(
            nn.AvgPool2d(kernel_size=3, stride=stride, padding=1),
            nn.Conv2d(in_ch, b4, kernel_size=1, stride=1, bias=False),
            nn.BatchNorm2d(b4),
            nn.ReLU(),
        )

        self.se = SEBlock(out_ch, reduction=8)

    def forward(self, x):
        y = torch.cat([self.branch1(x), self.branch2(x), self.branch3(x), self.branch4(x)], dim=1)
        return self.se(y)

class CustomCNN(nn.Module):
    def __init__(self, num_classes=200, dropout=0.4):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Conv2d(3, 48, kernel_size=5, stride=2, padding=2, bias=False),
            nn.BatchNorm2d(48),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=3, stride=2, padding=1),
        )

        self.block1 = MultiBranchBlock(48, 128, stride=1)
        self.block2 = MultiBranchBlock(128, 192, stride=2)
        self.block3 = MultiBranchBlock(192, 256, stride=2)
        self.block4 = MultiBranchBlock(256, 320, stride=2)

        self.mix = nn.Sequential(
            nn.Conv2d(320, 320, kernel_size=1, bias=False),
            nn.BatchNorm2d(320),
            nn.ReLU()
        )

        self.pool = nn.AdaptiveAvgPool2d((1, 1))
        self.drop = nn.Dropout(dropout)
        self.fc = nn.Linear(320, num_classes)

    def forward(self, x):
        x = self.stem(x)
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        x = self.mix(x)
        x = self.pool(x)
        x = torch.flatten(x, 1)
        x = self.drop(x)
        return self.fc(x)

def train_custom_model():
    model = CustomCNN(num_classes=num_classes, dropout=0.4).to(device)

    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
    optimizer = optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-3)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="max", patience=3, factor=0.2)
    scaler = torch.cuda.amp.GradScaler(enabled=use_amp)

    best_val_acc = 0.0
    best_path = "/content/best_custom_model_cub.pth"

    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}

    epochs = 50
    print("\nCustomCNN Training")
    for epoch in range(1, epochs + 1):
        tr_loss, tr_acc = train_one_epoch(model, custom_train_loader, optimizer, criterion, scaler)
        va_loss, va_acc = eval_one_epoch(model, custom_val_loader, criterion)
        scheduler.step(va_acc)

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(va_loss)
        history["val_acc"].append(va_acc)

        print(f"Epoch {epoch:02d}/{epochs} | train acc {tr_acc:.4f} | val acc {va_acc:.4f}")

        if va_acc > best_val_acc:
            best_val_acc = va_acc
            torch.save({"model_state": model.state_dict(),
                        "class_to_idx": custom_train_ds.class_to_idx,
                        "val_acc": best_val_acc}, best_path)
            print(f"Saved new best CustomCNN: val_acc={best_val_acc:.4f}")

    print("Best CustomCNN val acc:", best_val_acc)
    print("Best CustomCNN ckpt:", best_path)

    plot_learning_curves(history, title_prefix="Custom CNN")

    return best_path

# Evaluation
def load_resnet_from_ckpt(path):
    ckpt = torch.load(path, map_location=device)
    model = build_resnet50(num_classes, weights=None)
    model.load_state_dict(ckpt["model_state"])
    model.eval()
    return model

def load_custom_from_ckpt(path):
    ckpt = torch.load(path, map_location=device)
    model = CustomCNN(num_classes=num_classes, dropout=0.4).to(device)
    model.load_state_dict(ckpt["model_state"])
    model.eval()
    return model

def evaluate_and_report(model, loader, title, class_names):
    y_true, y_pred = predict_all(model, loader)
    m = compute_metrics(y_true, y_pred)

    print("\n" + "="*70)
    print(title)
    print(f"TEST Accuracy : {m['accuracy']:.4f}")
    print(f"Precision     : {m['precision']:.4f} (macro)")
    print(f"Recall        : {m['recall']:.4f} (macro)")
    print(f"F1            : {m['f1']:.4f} (macro)")
    print("="*70)

    print("\nClassification report:")
    print(classification_report(y_true, y_pred, target_names=class_names, digits=4, zero_division=0))
    plot_confusion_matrix(y_true, y_pred, title=f"{title} - Confusion Matrix")

# Test
best_resnet_ckpt = finetune_resnet50()
best_custom_ckpt = train_custom_model()

# Model 1
resnet_model = load_resnet_from_ckpt(best_resnet_ckpt)
evaluate_and_report(resnet_model, std_test_loader,
                    title="STANDARD MODEL: Pretrained ResNet50 (Test set)",
                    class_names=std_test_ds.classes)

# Model 2
custom_model = load_custom_from_ckpt(best_custom_ckpt)
evaluate_and_report(custom_model, custom_test_loader,
                    title="CUSTOM MODEL: CustomCNN (scratch, Test set)",
                    class_names=custom_test_ds.classes)

print("\nDone. Checkpoints saved at:")
print(" -", best_resnet_ckpt)
print(" -", best_custom_ckpt)
```
